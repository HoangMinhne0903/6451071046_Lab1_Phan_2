# Phần 2 - Bài 2 (Trang 37-38): Luật kết hợp - Online Retail Dataset


```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# Cài đặt thư viện cần thiết

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
from mlxtend.preprocessing import TransactionEncoder

# 1. ĐỌC DỮ LIỆU
df = pd.read_excel("Online Retail.xlsx", dtype={"InvoiceNo": str})
print("Kích thước dữ liệu gốc:", df.shape)
print("\n5 dòng đầu:")
print(df.head())
print("\nThông tin:")
print(df.info())
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 2. TIỀN XỬ LÝ
# Cắt ký tự thừa ở cột Description
df["Description"] = df["Description"].str.strip()

# Xóa các dòng không có InvoiceNo
df.dropna(subset=["InvoiceNo"], inplace=True)

# Chuyển InvoiceNo về kiểu chuỗi
df["InvoiceNo"] = df["InvoiceNo"].astype(str)

print("Sau tiền xử lý bước 2:")
print("Kích thước:", df.shape)
print(df.head())
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 3. XÓA HÓA ĐƠN TÍN DỤNG (InvoiceNo bắt đầu bằng 'C')
df = df[~df["InvoiceNo"].str.contains("C")]
print("Sau khi xóa hóa đơn tín dụng:")
print("Kích thước:", df.shape)
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 4. THỐNG KÊ SỐ DÒNG THEO QUỐC GIA
print("=" * 50)
print("Thống kê số dòng dữ liệu theo Quốc gia:")
print("=" * 50)
country_counts = df.groupby("Country")["InvoiceNo"].count().sort_values(ascending=False)
print(country_counts)
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 5. LẤY DỮ LIỆU UK VÀ GOM NHÓM
basket_uk = (df[df["Country"] == "United Kingdom"]
             .groupby(["InvoiceNo", "Description"])["Quantity"]
             .sum().unstack().reset_index().fillna(0)
             .set_index("InvoiceNo"))

print("Dữ liệu UK sau gom nhóm:")
print("Kích thước:", basket_uk.shape)
print(basket_uk.head())
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 6. CHUYỂN VỀ HOT ENCODING (>0 -> 1, <=0 -> 0)
def encode_units(x):
    return 1 if x >= 1 else 0

basket_uk_hot = basket_uk.applymap(encode_units)
print("Hot encoding:")
print(basket_uk_hot.head())
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 7. CHUYỂN VỀ ONE-HOT ENCODING (dạng boolean True/False)
basket_uk_bool = basket_uk_hot.astype(bool)
print("One-hot encoding (bool):")
print(basket_uk_bool.head())
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 8. XÓA CỘT POSTAGE
if "POSTAGE" in basket_uk_bool.columns:
    basket_uk_bool.drop(columns=["POSTAGE"], inplace=True)
    print("Đã xóa cột POSTAGE")
else:
    print("Không tìm thấy cột POSTAGE")

print("Kích thước sau xóa POSTAGE:", basket_uk_bool.shape)
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 9. TÌM TẬP PHỔ BIẾN BẰNG APRIORI (min_sup = 3%)
print("=" * 60)
print("APRIORI - min_support = 3%")
print("=" * 60)

frequent_items_apriori = apriori(basket_uk_bool, min_support=0.03, use_colnames=True)
frequent_items_apriori = frequent_items_apriori.sort_values("support", ascending=False)
print(f"Số tập phổ biến tìm được: {len(frequent_items_apriori)}")
print("\nTop 20 tập phổ biến:")
print(frequent_items_apriori.head(20))
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 10. TẠO LUẬT KẾT HỢP (min_conf = 50%)
print("=" * 60)
print("LUẬT KẾT HỢP - min_confidence = 50%")
print("=" * 60)

rules = association_rules(frequent_items_apriori, metric="confidence", min_threshold=0.5)
rules = rules.sort_values(["confidence", "lift"], ascending=[False, False])
print(f"Số luật kết hợp tìm được: {len(rules)}")
print("\nTop 20 luật kết hợp:")
print(rules[["antecedents", "consequents", "support", "confidence", "lift"]].head(20))
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 11. VẼ SCATTER PLOT - CONFIDENCE vs SUPPORT
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Support vs Confidence
sc = axes[0].scatter(rules["support"], rules["confidence"],
                     c=rules["lift"], cmap="coolwarm", alpha=0.7, s=30)
axes[0].set_xlabel("Support")
axes[0].set_ylabel("Confidence")
axes[0].set_title("Support vs Confidence (màu = Lift)")
plt.colorbar(sc, ax=axes[0], label="Lift")
axes[0].grid(True, linestyle="--", alpha=0.4)

# Support vs Lift
sc2 = axes[1].scatter(rules["support"], rules["lift"],
                      c=rules["confidence"], cmap="viridis", alpha=0.7, s=30)
axes[1].set_xlabel("Support")
axes[1].set_ylabel("Lift")
axes[1].set_title("Support vs Lift (màu = Confidence)")
plt.colorbar(sc2, ax=axes[1], label="Confidence")
axes[1].grid(True, linestyle="--", alpha=0.4)

plt.suptitle("Phân tích các luật kết hợp - Online Retail UK", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

# 12. FP-GROWTH VÀ SO SÁNH VỚI APRIORI
import time

print("=" * 60)
print("FP-GROWTH - min_support = 3%, min_confidence = 50%")
print("=" * 60)

start_apriori = time.time()
frequent_items_apriori2 = apriori(basket_uk_bool, min_support=0.03, use_colnames=True)
rules_apriori = association_rules(frequent_items_apriori2, metric="confidence", min_threshold=0.5)
time_apriori = time.time() - start_apriori

start_fp = time.time()
frequent_items_fp = fpgrowth(basket_uk_bool, min_support=0.03, use_colnames=True)
rules_fp = association_rules(frequent_items_fp, metric="confidence", min_threshold=0.5)
time_fp = time.time() - start_fp

print(f"\nApriori:")
print(f"  - Thời gian: {time_apriori:.4f} giây")
print(f"  - Số tập phổ biến: {len(frequent_items_apriori2)}")
print(f"  - Số luật kết hợp: {len(rules_apriori)}")

print(f"\nFP-Growth:")
print(f"  - Thời gian: {time_fp:.4f} giây")
print(f"  - Số tập phổ biến: {len(frequent_items_fp)}")
print(f"  - Số luật kết hợp: {len(rules_fp)}")

print(f"\n=> FP-Growth nhanh hơn Apriori {time_apriori/time_fp:.2f} lần")
print("=> Cả hai thuật toán cho kết quả giống nhau (cùng số tập phổ biến và luật)")
```
